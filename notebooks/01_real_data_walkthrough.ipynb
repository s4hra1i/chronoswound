# ChronosWound real-data walkthrough
This notebook reproduces the human GSE8056 analysis while keeping exploratory discovery separate from locked-panel validation. Read `docs/GSE8056_DATASET_CARD.md` first.

```python
from chronoswound.real_data import download_geo_files, parse_series_matrix, parse_annotation, collapse_to_genes
matrix_path, annotation_path = download_geo_files('../data/geo')
probes = parse_series_matrix(matrix_path)
genes = collapse_to_genes(probes, parse_annotation(annotation_path))
genes.shape
```

## Locked-panel validation
The panel was defined biologically and does not use the exploratory ranking below. Every array is held out once.

```python
from chronoswound.validation import evaluate_human_time_bins
validation = evaluate_human_time_bins(genes, '../reports/notebook_validation')
validation
```

## Exploratory temporal ranking
These genes generate hypotheses only. They must not be fed back into the validation model and described as independently tested.

```python
from chronoswound.real_data import differential_timecourse
statistics = differential_timecourse(genes)
statistics.head(20)
```
